```python
import re
import csv

def extract_text_between_tags(text, start_tag, end_tag=None):
    """Extract text content between HTML tags."""
    if end_tag:
        pattern = f'{re.escape(start_tag)}(.*?){re.escape(end_tag)}'
    else:
        pattern = f'{re.escape(start_tag)}(.*?)(?=</)'
    matches = re.findall(pattern, text, re.DOTALL)
    return matches[0] if matches else ""

def clean_html_text(text):
    """Remove HTML tags and clean up text."""
    # Remove HTML tags
    text = re.sub(r'<[^>]+>', '', text)
    # Decode HTML entities
    text = text.replace('&amp;', '&')
    text = text.replace('&quot;', '"')
    text = text.replace('&#039;', "'")
    text = text.replace('&lt;', '<')
    text = text.replace('&gt;', '>')
    # Clean up whitespace
    text = ' '.join(text.split())
    return text.strip()

def extract_table_data(html_content, facility_type):
    """Extract data from a table section."""
    rows = []
    
    # Find all table rows
    tr_pattern = r'<tr>(.*?)</tr>'
    table_rows = re.findall(tr_pattern, html_content, re.DOTALL)
    
    for row in table_rows:
        # Skip header rows
        if '<th' in row:
            continue
            
        # Extract table cells
        td_pattern = r'<td[^>]*>(.*?)</td>'
        cells = re.findall(td_pattern, row, re.DOTALL)
        
        if len(cells) >= 5:
            # Extract text from each cell
            city = clean_html_text(cells[0])
            region = clean_html_text(cells[1])
            oaci = clean_html_text(cells[2])
            iata = clean_html_text(cells[3])
            facility_name = clean_html_text(cells[4])
            
            # Only add if we have meaningful data
            if city and facility_name:
                rows.append({
                    'City_or_Location': city,
                    'Region': region,
                    'OACI': oaci,
                    'IATA': iata,
                    'Facility_Name': facility_name,
                    'Facility_Type': facility_type
                })
    
    return rows

def main():
    # Read the HTML file
    input_file = 'datos.txt'
    with open(input_file, 'r', encoding='utf-8') as f:
        content = f.read()
    
    all_data = []
    
    # Define sections with their facility types
    sections = [
        {
            'name': 'Aeropuertos (con vuelos comerciales)',
            'start': '<h3 id="Aeropuertos_(con_vuelos_comerciales)">',
            'end': '<h3 id="Bases_aéreas_(con_fuerza_aérea_o_militar)">'
        },
        {
            'name': 'Bases aéreas (con fuerza aérea o militar)',
            'start': '<h3 id="Bases_aéreas_(con_fuerza_aérea_o_militar)">',
            'end': '<h3 id="Aeródromos">'
        },
        {
            'name': 'Aeródromos',
            'start': '<h3 id="Aeródromos">',
            'end': '<h3 id="Helipuertos">'
        },
        {
            'name': 'Helipuertos',
            'start': '<h3 id="Helipuertos">',
            'end': '<div class="mw-heading mw-heading2"><h2 id="Véase_también">'
        }
    ]
    
    # Extract data from each section
    for section in sections:
        start_idx = content.find(section['start'])
        end_idx = content.find(section['end'])
        
        if start_idx != -1:
            if end_idx != -1:
                section_content = content[start_idx:end_idx]
            else:
                # For the last section, search for references section
                ref_idx = content.find('<div class="mw-heading mw-heading2"><h2 id="Referencias">')
                if ref_idx != -1:
                    section_content = content[start_idx:ref_idx]
                else:
                    section_content = content[start_idx:]
            
            rows = extract_table_data(section_content, section['name'])
            all_data.extend(rows)
            print(f"Extracted {len(rows)} entries from {section['name']}")
    
    # Write to CSV
    if all_data:
        with open('chile_airports.csv', 'w', newline='', encoding='utf-8') as csvfile:
            fieldnames = ['City_or_Location', 'Region', 'OACI', 'IATA', 'Facility_Name', 'Facility_Type']
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            
            writer.writeheader()
            writer.writerows(all_data)
        
        print(f"\nTotal entries extracted: {len(all_data)}")
        print("Data saved to chile_airports.csv")
    else:
        print("No data extracted!")

if __name__ == "__main__":
    main()
```
